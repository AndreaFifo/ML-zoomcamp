```python
def vector_vector_mult(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0

    for i in range(n):
        result = result + u[i] * v[i]

    return result
```

```python
b = np.arange(4)
b
```

```python
vector_vector_mult(a, b)
```

```python
U = np.array([
    [2, 4, 5, 6],
     [1, 2, 1, 2],
     [3, 1, 2, 1],
 ])
```

```python
def matrix_vector_mult(U, v):
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]

    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_mult(U[i], v)

    return result
```

```python
matrix_vector_mult(U, b)
```

```python
U.dot(b)
```
